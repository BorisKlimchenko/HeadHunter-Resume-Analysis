* Женщина , 21 год , родилась 13 января 2000
Как вы понимаете, нам необходимо выделить каждый параметр в отдельный столбец.

Создайте два новых признака **"Пол"** и **"Возраст"**. При этом важно учесть:
* Признак пола должен иметь 2 уникальных строковых значения: 'М' - мужчина, 'Ж' - женщина. 
* Признак возраста должен быть представлен целыми числами.

Выполните преобразование, ответьте на контрольные вопросы и удалите признак **"Пол, возраст"** из таблицы.

*Совет: обратите внимание на структуру текста в столбце, в части на то, как разделены параметры пола, возраста и даты рождения между собой - символом ' , '. 
Гарантируется, что структура одинакова для всех строк в таблице. Вы можете воспользоваться этим.*


```python
# 1. Функция для извлечения пола
def get_gender(text):
    if 'Мужчина' in text:
        return 'М'
    else:
        return 'Ж'

# 2. Функция для извлечения возраста
def get_age(text):
    # Разрезаем текст по запятой и берем второй кусок (индекс 1)
    age_chunk = text.split(',')[1]
    # Очищаем от лишних пробелов по краям (.strip), 
    # режем по пробелу и берем первое слово — это всегда цифра
    age_number = age_chunk.strip().split(' ')[0]
    return int(age_number)

# 3. Применяем функции и создаем два новых столбца
hh_data['Пол'] = hh_data['Пол, возраст'].apply(get_gender)
hh_data['Возраст'] = hh_data['Пол, возраст'].apply(get_age)

# 4. Удаляем старый мусорный столбец
hh_data = hh_data.drop('Пол, возраст', axis=1)

# Проверяем результат
display(hh_data[['Пол', 'Возраст']].head())
```

3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
def get_experience(text):
    # 1. Проверяем на пропуски (если значение пустое или равно "Не указано")
    if text is np.nan or text == 'Не указано':
        return np.nan
    
    # 2. Разрезаем текст по пробелам и берем только первые 7 слов
    words = text.split(' ')[:7]
    
    years = 0
    months = 0
    
    # 3. Списки ключевых слов для поиска
    year_words = ['год', 'года', 'лет']
    month_words = ['месяц', 'месяца', 'месяцев']
    
    # 4. Проходимся циклом по каждому слову
    for i in range(len(words)):
        # Если слово означает года, берем цифру перед ним (индекс i-1)
        if words[i] in year_words:
            years = int(words[i-1])
        # Если слово означает месяцы, берем цифру перед ним
        if words[i] in month_words:
            months = int(words[i-1])
            
    # 5. Возвращаем суммарный опыт в месяцах
    return years * 12 + months

# Применяем функцию к столбцу и создаем новый
hh_data['Опыт работы (месяц)'] = hh_data['Опыт работы'].apply(get_experience)

# Удаляем старый текстовый столбец
hh_data = hh_data.drop('Опыт работы', axis=1)

# Ответ на контрольный вопрос (в этом задании обычно просят медиану)
median_exp = hh_data['Опыт работы (месяц)'].median()
print(f"Медианный опыт работы: {median_exp} месяцев")
```

    * … , готов к переезду , …

    Еще один важный факт: при выгрузки данных у некоторых соискателей "потерялась" информация о готовности к командировкам. Давайте по умолчанию будем считать, что такие соискатели не готовы к командировкам.
    
Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
# Список городов из задания
million_cities = ['Новосибирск', 'Екатеринбург', 'Нижний Новгород', 'Казань', 'Челябинск', 'Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж', 'Волгоград']

# 1. Функция для города
def get_city(text):
    # Город всегда стоит до первой запятой
    city = text.split(',')[0].strip()
    if city == 'Москва':
        return 'Москва'
    elif city == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    elif city in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'

# 2. Функция для переезда
def get_relocation(text):
    # Ищем маркеры отказа. Если их нет — значит готов (или хочет)
    if 'не готов к переезду' in text or 'не готова к переезду' in text:
        return False
    return True

# 3. Функция для командировок
def get_business_trips(text):
    # Если слово вообще не упоминается, считаем False (по условию)
    if 'командировка' in text:
        if 'не готов к командировкам' in text or 'не готова к командировкам' in text:
            return False
        return True # Готов к редким или обычным командировкам
    return False

# 4. Применяем функции к столбцу
hh_data['Город'] = hh_data['Город, переезд, командировки'].apply(get_city)
hh_data['Готовность к переезду'] = hh_data['Город, переезд, командировки'].apply(get_relocation)
hh_data['Готовность к командировкам'] = hh_data['Город, переезд, командировки'].apply(get_business_trips)

# 5. Удаляем старый столбец
hh_data = hh_data.drop('Город, переезд, командировки', axis=1)
```

```python
# Считаем долю соискателей, живущих в Санкт-Петербурге (в процентах)
spb_percent = (hh_data['Город'] == 'Санкт-Петербург').mean() * 100
print(f"В Санкт-Петербурге живут: {round(spb_percent)}%")

# Считаем долю соискателей, готовых одновременно к переездам и командировкам
# Используем символ & для проверки выполнения двух условий сразу
ready_both_percent = (hh_data['Готовность к переезду'] & hh_data['Готовность к командировкам']).mean() * 100
print(f"Готовы и к переездам, и к командировкам: {round(ready_both_percent)}%")
```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Сейчас признаки представляют собой набор категорий желаемой занятости (полная занятость, частичная занятость, проектная работа, волонтерство, стажировка) и желаемого графика работы (полный день, сменный график, гибкий график, удаленная работа, вахтовый метод).
На сайте hh.ru соискатель может указывать различные комбинации данных категорий, например:
* полная занятость, частичная занятость
* частичная занятость, проектная работа, волонтерство
* полный день, удаленная работа
* вахтовый метод, гибкий график, удаленная работа, полная занятость

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>
Выполните данное преобразование для признаков "Занятость" и "График", ответьте на контрольные вопросы, после чего удалите их из таблицы

```python
# 1. Задаем списки всех возможных категорий из задания
employments = ['полная занятость', 'частичная занятость', 'проектная работа', 'волонтерство', 'стажировка']
schedules = ['полный день', 'сменный график', 'гибкий график', 'удаленная работа', 'вахтовый метод']

# 2. Создаем столбцы-мигалки для занятости
for emp in employments:
    # Метод str.contains() проверяет, есть ли искомое слово в ячейке
    hh_data[emp] = hh_data['Занятость'].str.contains(emp)

# 3. Создаем столбцы-мигалки для графика
for sch in schedules:
    hh_data[sch] = hh_data['График'].str.contains(sch)

# 4. Удаляем старые текстовые столбцы, так как они нам больше не нужны
hh_data = hh_data.drop(['Занятость', 'График'], axis=1)
```

```python
# Сколько людей ищут проектную работу и волонтерство
# Используем знак & (И) и считаем сумму совпадений (.sum())
proj_and_vol = (hh_data['проектная работа'] & hh_data['волонтерство']).sum()
print(f"Проектная работа и волонтерство: {proj_and_vol} человек")

# Сколько людей хотят работать вахтовым методом и с гибким графиком
vahta_and_flex = (hh_data['вахтовый метод'] & hh_data['гибкий график']).sum()
print(f"Вахтовый метод и гибкий график: {vahta_and_flex} человек")
```

Осталось только понять, откуда брать дату, по которой определяется курс? А вот же она - в признаке **"Обновление резюме"**, в нем содержится дата и время, когда соискатель выложил текущий вариант своего резюме. Нас интересует только дата, по ней бы и будем сопоставлять курсы валют.

Теперь у нас есть вся необходимая информация для того, чтобы создать признак "ЗП (руб)" - заработная плата в рублях.

После ответа на контрольные вопросы удалите исходный столбец заработной платы "ЗП" и все промежуточные столбцы, если вы их создавали.

Итак, давайте обсудим возможный алгоритм преобразования: 
1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
# 1. Загружаем курсы валют 
rates_data = pd.read_csv('data/ExchangeRates.csv')

# 2. Переводим даты в обеих таблицах в стандартный формат datetime, 
# а затем отсекаем время, оставляя только саму дату (.dt.date)
rates_data['date'] = pd.to_datetime(rates_data['date'], dayfirst=True).dt.date

# Для основной таблицы создаем временный столбец с датой, чтобы по нему объединять
hh_data['temp_date'] = pd.to_datetime(hh_data['Обновление резюме'], dayfirst=True).dt.date

# 3. Вытаскиваем сумму и валюту из строки "ЗП"
# Берем первый элемент до пробела и превращаем в число с плавающей точкой
hh_data['ЗП (сумма)'] = hh_data['ЗП'].apply(lambda x: float(x.split(' ')[0]))
# Берем второй элемент после пробела (саму валюту)
hh_data['ЗП (валюта)'] = hh_data['ЗП'].apply(lambda x: x.split(' ')[1])

# 4. Создаем словарь-переводчик для валют (по таблице из задания)
currency_dict = {
    'руб.': 'RUB', 'бел.руб.': 'BYN', 'KZT': 'KZT', 'EUR': 'EUR',
    'USD': 'USD', 'грн.': 'UAH', 'сум': 'UZS', 'KGS': 'KGS', 'AZN': 'AZN'
}
# Заменяем русские названия на международные коды ISO
hh_data['ЗП (валюта)'] = hh_data['ЗП (валюта)'].map(currency_dict)

# 5. Склеиваем таблицы с помощью метода merge
# how='left' гарантирует, что мы не потеряем резюме с рублями, которых нет в таблице курсов
merged_data = hh_data.merge(
    rates_data,
    left_on=['ЗП (валюта)', 'temp_date'],
    right_on=['currency', 'date'],
    how='left'
)

# 6. Заполняем пустоты (NaN) для рублевых зарплат
# Раз рублей в таблице курсов нет, их курс и пропорция равны 1
merged_data['close'] = merged_data['close'].fillna(1)
merged_data['proportion'] = merged_data['proportion'].fillna(1)

# 7. Считаем итоговую зарплату по формуле из задания
hh_data['ЗП (руб)'] = hh_data['ЗП (сумма)'] * (merged_data['close'] / merged_data['proportion'])

# 8. Тотальная зачистка: удаляем всё лишнее
hh_data = hh_data.drop(['ЗП', 'ЗП (сумма)', 'ЗП (валюта)', 'temp_date'], axis=1)
```

```python
# Считаем медиану, переводим в тысячи и округляем
median_salary = round(hh_data['ЗП (руб)'].median() / 1000)
print(f"Медианная ЗП: {median_salary}")
```

**Методологическая рефлексия: Изоляция деструктивных операций и управление памятью**

В процессе работы над очисткой массива я столкнулся с важным архитектурным нюансом среды Jupyter Notebook. Первичный анализ, при котором я использовал связку методов `hh_data.duplicated().sum()`, выполнял исключительно диагностическую функцию: алгоритм подсчитывал совпадения, но не вносил физических изменений в структуру датафрейма в оперативной памяти. Из-за этого последующие расчеты (например, поиск пропусков) производились на «грязных» данных, искажая результат.

Для решения этой проблемы я принял решение вынести метод физического удаления строк `hh_data = hh_data.drop_duplicates()` в отдельную, изолированную ячейку. 

Этот подход решает сразу две задачи:
1. **Программная:** Конструкция с переприсваиванием (`hh_data = ...`) гарантирует перезапись исходной переменной, навсегда очищая рабочий массив от 161 дубликата перед стартом следующих этапов.
2. **Архитектурная:** Изоляция необратимых (деструктивных) операций в отдельных ячейках является элементом оборонительного программирования. Это создает визуальный «рубильник» в блокноте, предотвращая случайное повторное выполнение мутирующего кода при перезапусках ядра и фиксируя четкую границу между этапом поиска дубликатов и этапом инвентаризации пропусков.

3. Итак, у нас есть пропуски в 3ех столбцах: **"Опыт работы (месяц)"**, **"Последнее/нынешнее место работы"**, **"Последняя/нынешняя должность"**. Поступим следующим образом: удалите строки, где есть пропуск в столбцах с местом работы и должностью. Пропуски в столбце с опытом работы заполните **медианным** значением.

```python
# 1. Ампутируем строки с пропусками в критически важных текстовых столбцах
# Параметр subset строго указывает алгоритму, в каких колонках искать "дыры"
hh_data = hh_data.dropna(subset=['Последнее/нынешнее место работы', 'Последняя/нынешняя должность'])

# 2. Вычисляем робастное медианное значение для стажа работы
median_experience = hh_data['Опыт работы (месяц)'].median()

# 3. Заполняем пропуски в стаже найденной медианой
hh_data['Опыт работы (месяц)'] = hh_data['Опыт работы (месяц)'].fillna(median_experience)

# 4. Контрольная проверка ответа для платформы
resulting_mean = hh_data['Опыт работы (месяц)'].mean()
print(f"Результирующее среднее значение в столбце «Опыт работы (месяц)»: {round(resulting_mean)}")
```

4. Мы добрались до ликвидации выбросов. Сначала очистим данные вручную. Удалите резюме, в которых указана заработная плата либо выше 1 млн. рублей, либо ниже 1 тыс. рублей.

```python
# Выделяем целевой столбец в отдельную переменную (принцип DRY и стандарт PEP 8).
# Это избавляет систему от двойного поиска ключа в памяти и делает код визуально чистым.
salary = hh_data['ЗП (руб)']

# Создаем логическую маску аномалий: метку True получат зарплаты > 1 млн ИЛИ < 1 тыс.
outlier_mask = (salary > 1_000_000) | (salary < 1_000)

# Подсчитываем количество True (выбросов) для ввода ответа на платформу SkillFactory.
outliers_count = int(outlier_mask.sum())
print(f'Количество найденных выбросов: {outliers_count}')

# Фильтруем данные: символ ~ (тильда) инвертирует маску, оставляя только "нормальные" строки.
# Метод .copy() жестко выделяет новый блок памяти, защищая нас от ошибки SettingWithCopyWarning.
hh_data = hh_data.loc[~outlier_mask].copy()
```

**Архитектурная рефлексия: Эвристическая фильтрация и управление памятью**

Приступая к ликвидации выбросов по заработной плате, я уделил особое внимание не только бизнес-логике (отсечение сумм < 1k и > 1M рублей), но и чистоте самого кода (PEP 8) и безопасности работы с памятью.

Для повышения читаемости и соблюдения принципа DRY, целевой признак был предварительно изолирован в переменную `salary`. Механика очистки построена не на деструктивном методе `drop`, а на логической инверсии (оператор `~`): датафрейм отфильтровывает и оставляет только те строки, которые НЕ попадают под условия аномалии. 

Ключевым архитектурным решением здесь является вызов метода `.copy()` в финале. При слайсинге через `.loc` Pandas склонен возвращать view (представление) исходного массива, что в дальнейшем может привести к конфликтам памяти (`SettingWithCopyWarning`). Явное копирование принудительно выделяет новый блок оперативной памяти, гарантируя, что дальнейшие трансформации будут производиться на абсолютно независимом и структурно чистом объекте. 

5. В процессе разведывательного анализа мы обнаружили резюме, в которых **опыт работы в годах превышал возраст соискателя**. Найдите такие резюме и удалите их из данных


```python
# Векторное сравнение признаков: находим логические аномалии. 
# Метку True получают строки, где стаж физически превышает возраст соискателя.
experience_age_mask = (hh_data['Опыт работы (лет)'] > hh_data['Возраст'])

# Подсчитываем аномалии, принудительно фиксируя результат как целое число (int) для вывода.
experience_age_count = int(experience_age_mask.sum())
print(f'Количество резюме с опытом больше возраста: {experience_age_count}')

# Инвертируем маску (~) — оставляем только адекватные анкеты.
# Выделяем чистый блок памяти через .copy() во избежание SettingWithCopyWarning.
hh_data = hh_data.loc[~experience_age_mask].copy()
```

Попробуйте построить распределение признака в **логарифмическом масштабе**. Добавьте к графику линии, отображающие **среднее и границы интервала метода трех сигм**. Напомним, сделать это можно с помощью метода axvline. Например, для построение линии среднего будет иметь вид:

`histplot.axvline(log_age.mean(), color='k', lw=2)`

В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Логарифмируем признак для приведения его к форме, близкой к нормальному распределению.
log_age = np.log(hh_data['Возраст'])
log_age_mean = log_age.mean()
log_age_std = log_age.std()

# 2. Визуализация распределения и границ метода сигм.
fig, ax = plt.subplots(figsize=(10, 5))
sns.histplot(log_age, bins=30, ax=ax)

ax.axvline(log_age_mean, color='black', linewidth=2, label='Среднее')
ax.axvline(log_age_mean - 3 * log_age_std, color='orange', linestyle='--', label='Среднее - 3σ')
ax.axvline(log_age_mean + 3 * log_age_std, color='red', linestyle='--', label='Среднее + 3σ')
# Визуализируем ту самую "послабленную" границу в 4 сигмы
ax.axvline(log_age_mean + 4 * log_age_std, color='purple', linestyle='-', label='Правая граница (+4σ)')

ax.set_title('Распределение логарифма возраста')
ax.set_xlabel('ln(Возраст)')
ax.set_ylabel('Количество резюме')
ax.legend()
plt.show()

# 3. Автоматическое определение направления асимметрии.
log_age_skew = log_age.skew()
if log_age_skew > 0:
    skew_direction = 'правосторонняя асимметрия'
elif log_age_skew < 0:
    skew_direction = 'левосторонняя асимметрия'
else:
    skew_direction = 'распределение практически симметрично'
print(f'Асимметрия log(Возраст): {log_age_skew:.3f} ({skew_direction})')

# 4. Расчет Z-оценки (z-score) и создание строгой двусторонней маски выбросов.
# Мы ищем резюме за пределами асимметричного коридора: строже слева (-3) и мягче справа (+4).
age_z_score = (log_age - log_age_mean) / log_age_std
age_outlier_mask = (age_z_score < -3) | (age_z_score > 4)

# 5. Изоляция выбросов для визуальной оценки перед ампутацией.
age_outliers = hh_data.loc[
    age_outlier_mask, 
    ['Возраст', 'Опыт работы (лет)', 'Город', 'Образование', 'ЗП (руб)']
].sort_values('Возраст')

print(f'Количество возрастных выбросов: {len(age_outliers)}')
display(age_outliers)

# 6. Физическое удаление аномалий через инверсию маски и безопасное копирование памяти.
hh_data = hh_data.loc[~age_outlier_mask].copy()
```
